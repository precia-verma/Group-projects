# 🏚️ Mansion Level 2: Adventure Game Development Blog

## 📅 Project Timeline: October 2025

Welcome to our development blog for **Mansion Level 2** - an exciting adventure game where players explore a haunted mansion and face spooky challenges!

---

## 🎮 Game Overview

Mansion Level 2 is an interactive browser-based game built with vanilla JavaScript and HTML5 Canvas. Players control a character using WASD controls to navigate through a haunted mansion environment, discovering secret areas like the mysterious cemetery.

### Key Features (For Now)

- **WASD Movement Controls**: Smooth character movement in all four directions
- **Interactive Environment**: Cemetery area with collision detection
- **Decision-Based Gameplay**: Players make choices that affect their journey
- **Atmospheric Design**: Haunted mansion background with spooky themes
- **Responsive Canvas**: Full-screen gameplay that adapts to window size

---

## 🛠️ Technical Implementation

### File Structure

Our game consists of two main files:
- `mansionLevel2.js` - Core game logic and mechanics
- `mansion2.html` - Game container and initialization

### Core Components

#### 1. **Player Character System**
```javascript
const player = {
  x: 50,
  y: 550,
  width: 30,
  height: 30,
  speed: 3,
  color: '#ff6b6b'
};
```
- Circular character design with white outline
- Starting position at coordinates (50, 550)
- Movement speed of 3 pixels per frame
- Direction indicator dot

#### 2. **WASD Controls**
We implemented a keyboard state system that tracks which keys are pressed:
- **W** - Move Up
- **A** - Move Left  
- **S** - Move Down
- **D** - Move Right

The controls are displayed on-screen with visual feedback showing active keys in green.

#### 3. **Cemetery Area**
```javascript
const cemetery = {
  x: 500,
  y: 200,
  width: 700,
  height: 450
};
```
A special collision zone where players encounter a prompt.

#### 4. **Interactive Prompt System**
When players enter the cemetery, they're presented with a dialogue box:
- **"Do you dare to enter?"**
- Two buttons: "Sure!" and "Not Ready"
- Keyboard shortcuts: Y or N
- Darkened background for focus

#### 5. **Transition Screen**
After accepting the cemetery challenge:
- Dramatic black screen fade-in
- Red glowing text: "PREPARE TO FACE A CHALLENGE"
- Shadow effects for atmospheric impact
- 3-second duration before next phase (another new website)

## 💻 Code Highlights

### Collision Detection
```javascript
function checkCemeteryCollision() {
  const playerCenterX = player.x + player.width / 2;
  const playerCenterY = player.y + player.height / 2;

  const isInCemetery = 
    playerCenterX >= cemetery.x &&
    playerCenterX <= cemetery.x + cemetery.width &&
    playerCenterY >= cemetery.y &&
    playerCenterY <= cemetery.y + cemetery.height;
}
```
Uses player center point to determine if they've entered the cemetery zone.

### Game Loop
Continuous rendering at 60 FPS:
1. Clear canvas
2. Draw background
3. Draw cemetery boundary
4. Draw player
5. Draw UI controls
6. Draw prompts (if active)
7. Request next animation frame

## 🎯 Development Challenges & Solutions

### Challenge 1: Canvas Focus
**Problem**: Canvas wasn't receiving keyboard input initially.  
**Solution**: Added `tabIndex` attribute and auto-focus functionality with a 100ms delay.

### Challenge 2: Smooth Movement
**Problem**: Character movement felt choppy.  
**Solution**: Implemented frame-based game loop using `requestAnimationFrame` instead of `setInterval`.

### Challenge 3: Background Loading
**Problem**: Background image path issues across different environments.  
**Solution**: Created fallback system with multiple potential paths and error handling.

### Challenge 4: Button Detection
**Problem**: Determining if player clicked on prompt buttons.  
**Solution**: Stored button coordinates and implemented click position checking with mouse event handling.
