# 🏚️ Mansion Level 2: Adventure Game Development Blog

## 📅 Project Timeline: October 2025

Welcome to our development blog for **Mansion Level 2** - an exciting adventure game where players explore a haunted mansion with animated character sprites and face spooky challenges!

---

## 🎮 Game Overview

Mansion Level 2 is an interactive browser-based game built with vanilla JavaScript and HTML5 Canvas. Players control an animated character using WASD controls to navigate through a haunted mansion environment, discovering secret areas like the mysterious cemetery.

### Key Features

- **6-Frame Character Animation**: Smooth sprite-based character with walking animation
- **WASD Movement Controls**: Fluid character movement in all four directions
- **Responsive Design**: Cemetery boundaries and game elements scale proportionally to screen size
- **Interactive Environment**: Cemetery area with dynamic collision detection
- **Decision-Based Gameplay**: Players make choices that affect their journey
- **Atmospheric Design**: Haunted mansion background with spooky themes
- **Mobile & Desktop Support**: Works on both keyboard and touch controls
- **Full-Screen Gameplay**: Responsive canvas that adapts to any window size

## 🛠️ Technical Implementation

### File Structure

Our game consists of two main files:
- `mansionLevel2.js` - Core game logic, sprite system, and mechanics

### Core Components

#### 1. **Player Character System with Sprite Animation**
```javascript
const player = {
  x: 50,
  y: 550,
  width: 250,
  height: 250,
  speed: 3,
  color: '#ff6b6b',
  isMoving: false

```
- **6-frame sprite animation** system (character_frame_1.png through character_frame_6.png)
- Large 250x250 pixel character for better visibility
- Smooth animation at 12 frames per sprite change
- Fallback colored circle when sprites are loading
- Starting position at coordinates (50, 550)
- Movement speed of 3 pixels per frame

#### 2. **Sprite Loading System**

#### 3. **WASD Controls**
We implemented a keyboard state system that tracks which keys are pressed:
- **W** - Move Up
- **A** - Move Left  
- **S** - Move Down
- **D** - Move Right

The controls are displayed on-screen with visual feedback showing active keys in green.

**Animation System**:
- Cycles through 6 sprite frames when moving
- Resets to frame 0 when idle (standing still)
- Smooth transitions at 12 frame intervals

#### 4. **Proportional Cemetery Area**
```javascript
function updateCemeteryBounds() {
  cemetery.width = canvas.width * 0.5;    // 50% of screen width
  cemetery.height = canvas.height * 0.6;   // 60% of screen height
  cemetery.x = (canvas.width - cemetery.width) / 2;  // Centered
  cemetery.y = canvas.height * 0.15;       // 15% from top
}
```
A special collision zone that scales with screen size:
- **Width**: 50% of screen width
- **Height**: 60% of screen height
- **Position**: Centered horizontally, 15% from top
- **Responsive**: Auto-updates on window resize

#### 5. **Interactive Prompt System**
When players enter the cemetery, they're presented with a dialogue box:
- **"Do you dare to enter?"**
- Two buttons: "Sure!" and "Not Ready"
- Keyboard shortcuts: Y or N
- Click detection with hover effects
- Darkened background for focus

#### 6. **Transition Screen**
After accepting the cemetery challenge:
- Dramatic black screen fade-in
- Red glowing text: "PREPARE TO FACE A CHALLENGE"
- Shadow effects for atmospheric impact
- 3-second duration before next phase (leads to new website/level)

## 💻 Code Highlights

### Sprite Animation Logic
```javascript
if (player.isMoving) {
  spriteAnimationCounter++;
  if (spriteAnimationCounter >= spriteAnimationSpeed) {
    spriteAnimationCounter = 0;
    currentSpriteIndex = (currentSpriteIndex + 1) % 6;
  }
} else {
  spriteAnimationCounter = 0;
  currentSpriteIndex = 0;  // Reset to idle frame
}
```
Smooth animation cycling through 6 frames when moving, with reset to idle when stopped.

### Responsive Cemetery Boundaries
```javascript
window.addEventListener('resize', () => {
  canvas.width = window.innerWidth;
  canvas.height = window.innerHeight;
  updateCemeteryBounds(); // Update proportionally on resize
});
```
Cemetery area scales with window size, maintaining proportions.

### Collision Detection
```javascript
function checkCemeteryCollision() {
  const playerCenterX = player.x + player.width / 2;
  const playerCenterY = player.y + player.height / 2;

  const isInCemetery = 
    playerCenterX >= cemetery.x &&
    playerCenterX <= cemetery.x + cemetery.width &&
    playerCenterY >= cemetery.y &&
    playerCenterY <= cemetery.y + cemetery.height;
}
```
Uses player center point to determine if they've entered the dynamically-sized cemetery zone.

### Game Loop
Continuous rendering at 60 FPS:
1. Clear canvas
2. Draw background
3. Draw cemetery boundary (proportional)
4. Draw player sprite (animated)
5. Draw UI controls
6. Draw prompts (if active)
7. Request next animation frame

## 🎯 Development Challenges & Solutions

### Challenge 1: Sprite Sheet vs Individual Files
**Problem**: Initially tried loading a sprite sheet that didn't exist.  
**Solution**: Refactored to load 6 individual sprite PNG files with proper error handling and fallback sprites.

### Challenge 2: Sprite Size Mismatch
**Problem**: 682x682 pixel sprites rendered at 120x120 showed only partial character.  
**Solution**: Increased render size to 250x250 pixels to display more of the character body while maintaining performance.

### Challenge 3: Merge Conflicts Breaking JavaScript
**Problem**: Git merge conflicts left conflict markers in the code, causing black screen.  
**Solution**: Implemented thorough conflict resolution and brace-counting validation to ensure syntax correctness.

### Challenge 4: Fixed Cemetery Boundaries
**Problem**: Cemetery area had fixed pixel coordinates that didn't adapt to screen size.  
**Solution**: Created `updateCemeteryBounds()` function that calculates proportional boundaries (50% width, 60% height) and updates on resize.

### Challenge 5: Canvas Focus
**Problem**: Canvas wasn't receiving keyboard input initially.  
**Solution**: Added `tabIndex` attribute and auto-focus functionality with a 100ms delay.

### Challenge 6: Smooth Animation
**Problem**: Character movement and sprite animation felt choppy.  
**Solution**: Implemented frame-based game loop using `requestAnimationFrame` with animation counter at 12 frames per sprite change.

### Challenge 7: Sprite Loading Reliability
**Problem**: Sprites sometimes wouldn't render even when loaded.  
**Solution**: Added `sprite.complete` and `sprite.naturalWidth > 0` validation before rendering, with fallback circle and "Loading sprites..." message.
