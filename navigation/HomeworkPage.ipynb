<h2>Key Concepts</h2>

<ul>
  <li>A boolean is a data type that has only two possible values: <strong>True</strong> and <strong>False</strong></li>
  <li>Booleans let programs make decisions based on conditions.</li>
  <li>Common uses of a Boolean include: determining if a number is positive, checking if a file exists, and controlling program flow with if/else.</li>
</ul>

<h3>Different types of Booleans</h3>
<ul>
  <li>Direct assignment: <code>is_sunny = True</code></li>
  <li>Comparisons: <code>is_adult = 18 >= 18</code> → evaluates to True</li>
</ul>

<h3>Examples of Boolean Operators</h3>
<ul>
  <li><code>and</code> – True if both conditions are True</li>
  <li><code>or</code> – True if at least one condition is True</li>
  <li><code>not</code> – Flips True to False (and vice versa)</li>
</ul>
