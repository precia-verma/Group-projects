### Key Concepts:

##### (-) A boolean is a data type that has only two possible values : True and False 

##### (-) Booleans let programs make decisions based on conditions.

##### (-) Common Uses of a Boolean include: determining if a number is positive, checking if a file exists, and controlling program flow with if/else.

##### Different types of Booleans:

##### (-) Direct assingment: is_sunny = True

##### (-) Comparisons: is_adult = 18 >= 18 → evaluates to True


##### Examples of Boolean Operators: 

##### - and - True if both conditions are True
##### - or -  True if at least one condition is True
##### - not - Flips True to False (and vice versa)
