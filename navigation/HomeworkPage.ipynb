<ul style="font-size:1.15em; color:#e3f2fd; font-weight:500; margin-bottom:18px;">
  <li><b style="color:#b3cde0;">Boolean</b> is a data type with only two possible values: <span style="color:#90caf9; font-weight:bold;">True</span> and <span style="color:#1976d2; font-weight:bold;">False</span>.</li>
  <li><b style="color:#b3cde0;">Booleans</b> let programs make decisions based on conditions.</li>
  <li>Common uses: <b style="color:#b3cde0;">is_positive</b> (number > 0), <b style="color:#b3cde0;">file_exists</b>, controlling flow with <b style="color:#b3cde0;">if/else</b>.</li>
</ul>

<h4 style="font-size:1.2em; color:#b3cde0; font-weight:700; margin-top:18px;">Types of Booleans</h4>

  <li><b>Direct assignment:</b> <span style="background:#1976d2; color:#e3f2fd; padding:2px 8px; border-radius:4px;">is_sunny = True</span></li>
  <li><b>Comparisons:</b> <span style="background:#90caf9; color:#0a2342; padding:2px 8px; border-radius:4px;">is_adult = 18 &gt;= 18</span> <span style="color:#90caf9; font-weight:bold;">True</span></li>
</ul>

<h4 style="font-size:1.2em; color:#b3cde0; font-weight:700; margin-top:18px;">Boolean Operators</h4>
<ul style="font-size:1.1em; color:#e3f2fd;">
  <li><b>and</b> <span style="color:#90caf9; font-weight:bold;">True</span> if <b>both</b> conditions are True</li>
  <li><b>or</b> <span style="color:#b3cde0; font-weight:bold;">True</span> if <b>at least one</b> condition is True</li>
  <li><b>not</b> <span style="color:#1976d2; font-weight:bold;">Flips</span> True to False (and vice versa)</li>
</ul>
